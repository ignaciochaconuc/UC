```python
def numero_a_binario(numero):

    signo = []
    bits_enteros = []
    bits_fraccionarios = []
    if numero == 0:
        return ["0"] * 32
    if numero < 0:
        signo.append("1")
    if numero > 0:
        signo.append("0")

    numero = abs(numero)
    parte_entera = int(numero)
    parte_fraccionaria = numero - parte_entera

    while parte_entera > 0:
        if parte_entera % 2 == 0:
            parte_entera = parte_entera // 2
            bits_enteros.append("0")
        else:
            parte_entera = parte_entera // 2
            bits_enteros.append("1")
    bits_enteros.reverse()

    i = 0
    while parte_fraccionaria != 0 and i < 150:
        parte_fraccionaria *= 2
        bit = int(parte_fraccionaria)
        bits_fraccionarios.append(str(bit))
        parte_fraccionaria -= bit
        i += 1

    if "1" not in bits_enteros:
        posicion_primer_uno = bits_fraccionarios.index("1")
        exponente = -(posicion_primer_uno + 1)
        mantisa = bits_fraccionarios[posicion_primer_uno + 1:]
    else:
        exponente = len(bits_enteros) - 1
        mantisa = bits_enteros[1:] + bits_fraccionarios

    exponente_sesgado = exponente + 127
    bits_exponente = []
    while exponente_sesgado > 0:
        if exponente_sesgado % 2 == 0:
            exponente_sesgado = exponente_sesgado // 2
            bits_exponente.append("0")
        else:
            exponente_sesgado = exponente_sesgado // 2
            bits_exponente.append("1")
    bits_exponente.reverse()
    bits_exponente = (["0"] * (8 - len(bits_exponente)) + bits_exponente)
    mantisa = (mantisa + ["0"] * 23)[:23]

    return signo + bits_exponente + mantisa
```

Esta función convierte un número real a su representación binaria de 32 bits (`float32`). Primero determina el bit de signo y separa las partes entera y fraccionaria para convertirlas a binario. Luego normaliza el número, calcula el exponente con sesgo 127 y completa la mantisa hasta 23 bits. Finalmente, retorna una lista formada por 1 bit de signo, 8 bits de exponente y 23 bits de mantisa.

```python
numero = 10.625

print(numero_a_binario(numero), len(numero_a_binario(numero)))
```

```python
def precision_computador():
    epsilon = 1.0
    bits = 0

    while 1.0 + epsilon / 2.0 != 1.0:
        epsilon = epsilon / 2.0
        bits += 1

    return epsilon, bits

epsilon, bits = precision_computador()

print("Épsilon de máquina:", epsilon)
print("Bits de la fracción:", bits)
```

El código lo que hace es buscar es la diferencia más pequeña que el computador alcanza a detectar como 2 números diferentes. Para esto, hace un ciclo que no para hasta la suma entre 1 y un número que disminuye a la mitad, sea igual a 1.

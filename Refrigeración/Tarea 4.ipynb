```python
import pandas as pd
import numpy as np

archivo = "330020_2025_Temperatura_.csv"
Tbal = 20

df = pd.read_csv(archivo, sep=";")

df["Instante"] = pd.to_datetime(df["Instante"])
df["Ts"] = pd.to_numeric(df["Ts"], errors="coerce")

inicio = "2025-06-21 00:00"
fin = "2025-09-20 23:00"

periodo = df[(df["Instante"] >= inicio) & (df["Instante"] <= fin)].copy()

periodo["GH"] = np.maximum(0, Tbal - periodo["Ts"])

GH_total = periodo["GH"].sum()

print("Horas usadas:", len(periodo))
print("GH total:", GH_total, "°C·h")
```

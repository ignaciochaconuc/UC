```python
import glob
import pandas as pd

# Toma TODOS los .xlsx de la carpeta donde corras este script.
# Si quieres, cambia "." por la ruta a tu carpeta, ej: r"C:\Users\tu_usuario\Downloads"
CARPETA = r"C:\UC\Refrigeración\Tarea 5"

archivos = glob.glob(f"{CARPETA}/*.xlsx")
print(f"Encontrados {len(archivos)} archivos")

dfs = []
for f in archivos:
    df = pd.read_excel(f, sheet_name="Export", parse_dates=["time"])
    dfs.append(df)

combinado = pd.concat(dfs, ignore_index=True)

# Quitar filas repetidas (por solapamiento) y ordenar por fecha/hora
combinado = combinado.drop_duplicates(subset="time").sort_values("time").reset_index(drop=True)

combinado.to_excel("2025_completo.xlsx", index=False)
print(f"Listo: {len(combinado)} filas guardadas en 2025_completo.xlsx")
print(f"Rango: {combinado['time'].min()} a {combinado['time'].max()}")
```
